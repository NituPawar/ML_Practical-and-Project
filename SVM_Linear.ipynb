```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("bill_authentication.csv")
data.shape
```

```python
data.head(10)
```

```python
X = data.drop('Class', axis=1)
Y = data['Class']
```

```python
print(X)
```

```python
print(Y)
```

```python
from sklearn.model_selection import train_test_split
Xtrain, Xtest, ytrain,ytest =train_test_split(X,Y,test_size =0.20,random_state=0)
```

```python
# Linear(Logistic) 
from sklearn.svm import SVC
svcl=SVC(kernel='linear')
svcl.fit(Xtrain, ytrain)
```

```python
y_pred=svcl.predict(Xtest)
```

```python
print(ytest)
```

```python
print(y_pred)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
print(classification_report(ytest,y_pred))
print(confusion_matrix(ytest,y_pred))
```

```python
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
```

```python
Xtrain, Xtest, ytrain,ytest =train_test_split(X,Y,test_size =0.20,random_state=0)
```

```python
linear_svm = SVC (kernel= 'linear')
linear_svm.fit(Xtrain, ytrain)
y_pred_linear = linear_svm.predict(Xtest)
print(f'Linear Kernel Accuracy: {accuracy_score(ytest, y_pred_linear)}')
print (classification_report(ytest, y_pred_linear))
```
